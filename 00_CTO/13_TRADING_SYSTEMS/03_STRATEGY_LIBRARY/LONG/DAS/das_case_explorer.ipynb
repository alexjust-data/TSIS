## Lectura DAS inicial

La busqueda empieza simulando el screener operativo, no filtrando DAS perfecto desde el inicio.

Filtro duro inicial:

```text
market_cap < 100M
session_volume >= 500000
0.5 <= price <= 20
```

Cada aparicion en screener se guarda como `Scanner_Appearance_Candidate`. Despues el notebook mide si hubo push, dip, rebreak y como fue la estructura.

La rotura del first push high tiene dos lecturas:

- puede ser una entrada candidata por rebreak/impulso;
- tambien puede activar la secuencia para estudiar dips posteriores como `DAS #1`, `DAS #2`, `DAS #3`.

La busqueda inicial etiqueta de forma aproximada el estado: `scanner_only`, `push_detected`, `push_and_dip`, `rebreak_confirmed` o `failed_before_rebreak`.
